### Exercise 2 (Easy)
Perform word analogy: king - man + woman = ?

```python
# king - man + woman = ?
result = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print(f"king - man + woman = {result[0][0]}")

# paris - france + spain = ?
result_geo = model.most_similar(positive=['paris', 'spain'], negative=['france'], topn=1)
print(f"paris - france + spain = {result_geo[0][0]}")
```

### Exercise 3 (Medium)
Train your own Word2Vec model on a custom corpus.

```python
from gensim.models import Word2Vec

corpus = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "ran", "in", "the", "park"],
    ["cats", "and", "dogs", "are", "pets"],
    ["the", "cat", "chased", "the", "dog"],
    ["pets", "need", "food", "and", "water"]
]

# Train Word2Vec model
model_custom = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, workers=4)

# Test custom model
print("Vector for 'cat':", model_custom.wv['cat'][:5]) # show first 5 dims
```

### Exercise 4 (Medium)
Build a simple neural network for text classification using embeddings.

```python
import torch
import torch.nn as nn

class TextClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.fc = nn.Linear(embed_dim, num_classes)
    
    def forward(self, x):
        # x shape: (batch_size, seq_len)
        embedded = self.embedding(x)  # (batch_size, seq_len, embed_dim)
        
        # Mean pooling across the sequence dimension (dim 1)
        pooled = torch.mean(embedded, dim=1) # (batch_size, embed_dim)
        
        # Final classification
        return self.fc(pooled)

# Test with dummy data (5 words, 10-dim embedding, 2 classes)
model_nn = TextClassifier(vocab_size=100, embed_dim=10, num_classes=2)
dummy_input = torch.randint(0, 100, (1, 5)) # 1 batch of 5 words
output = model_nn(dummy_input)
print("Classifier Output:", output)
```
